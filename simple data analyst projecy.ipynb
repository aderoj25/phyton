```python
#Segment wise total sales
total_sales_by_Segment = df.groupby("Segment")["Sales"].sum().reset_index()
total_sales_by_Segment['Sales']=total_sales_by_Segment['Sales'].round(2)
print(total_sales_by_Segment)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))
plt.bar(total_sales_by_Segment['Segment'],total_sales_by_Segment['Sales'])
plt.xlabel('Segment')
plt.ylabel('Total Sales')
plt.title('Segment Wise Total Sales')
plt.xticks(rotation=90)
plt.show()
```

```python
#Category wise average sales
average_sales_by_category = df.groupby("Category")["Sales"].mean().reset_index()
average_sales_by_category['Sales']= average_sales_by_category['Sales'].round(2)
print(average_sales_by_category)
```

```python
plt.figure(figsize=(8,6))
plt.bar(average_sales_by_category['Category'],average_sales_by_category['Sales'])
plt.xlabel('Category')
plt.ylabel('Average Sales')
plt.title('Category Wise Average Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Group by 'Region' and 'category' and sum 'sales'
category_region_sales = df.groupby(['Region', 'Category'])['Sales'].sum().reset_index()

#Create a clustered bar chart
plt.figure(figsize=(15, 8))
sns.barplot(x='Region', y='Sales', hue='Category', data=category_region_sales)
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.title('Category-wise Total Sales by Region')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
