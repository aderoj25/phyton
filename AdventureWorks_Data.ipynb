```python
#customer analysis
total_customers = customer["CustomerKey"].nunique()
print("Total customers:", total_customers)

average_income = customer["AnnualIncome"].mean()
print(average_income)

highest_income = customer["AnnualIncome"].max()
print(highest_income)

occupation_counts = customer["Occupation"].value_counts()
most_common_occupation = occupation_counts.idxmax()
print( most_common_occupation)

homeowners_count = (customer["HomeOwner"] == "Y").sum()
print(homeowners_count)
```

```python
#Product Analysis

total_products = product["ProductKey"].nunique()
print(total_products)

average_price = product["ProductPrice"].mean()
print(average_price)

highest_price = product["ProductPrice"].max()
print(highest_price)

average_cost = product["ProductCost"].mean()
print(average_cost)

color_counts = product["ProductColor"].value_counts()
most_common_color = color_counts.idxmax()
print(most_common_color)
```

```python
# Sales Analysis
# combine separate years into one single 'sales' dataFrame
sales = pd.concat([sales_2020, sales_2021, sales_2022])
```

```python
top_sales_year= sales["OrderDate"].dt.year.value_counts().idxmax()
print(top_sales_year)

top_quantity_year= sales.groupby("OrderDate")["OrderQuantity"].sum()
print(top_quantity_year)

max_quantity_2022 = sales_2022["OrderQuantity"].max()
print(max_quantity_2022)

unique_customers_2022 = sales_2022["CustomerKey"].nunique()
print(unique_customers_2022)

unique_products_2022 = sales_2022["ProductKey"].nunique()
print(unique_products_2022)



```
