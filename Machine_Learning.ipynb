```python
from google.colab import drive
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

drive.mount('/content/drive')
file_path = '/content/drive/MyDrive/Car Price.xlsx'

df = pd.read_excel(file_path)
df
```

```python
x = df[['year','km_driven','mileage(km/ltr/kg)', 'engine', 'max_power', 'seats']]
y = df['selling_price']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

```python
model = LinearRegression()

# Convert all columns in x to numeric, coercing errors to NaN
for col in x.columns:
    x[col] = pd.to_numeric(x[col], errors='coerce')

# Convert y to numeric, coercing errors to NaN
y = pd.to_numeric(y, errors='coerce')

# Fill NaN values with 0 in both x and y
x_cleaned = x.fillna(0)
y_cleaned = y.fillna(0)

model.fit(x_cleaned, y_cleaned)
```

```python
new_data = pd.DataFrame([[2017,10000,55.0,1250,73.5,4]], columns=['year', 'km_driven','mileage(km/ltr/kg)','engine','max_power','seats'])

prediction = model.predict(new_data)
print("Predicted Selling Price:", prediction)
```
